```python
import yt_dlp

def download_best_quality(url):
    try:
        # Opsi untuk mendownload video dan audio terbaik lalu digabungkan
        ydl_opts = {
            'format': 'bestvideo+bestaudio/best',  # Mengambil video dan audio terbaik lalu menggabungkannya
            'outtmpl': '%(title)s.%(ext)s',        # Menyimpan file dengan nama berdasarkan judul video
            'noplaylist': True,                    # Jangan mendownload playlist
            'merge_output_format': 'mp4',          # Gabungkan menjadi MP4
        }

        # Menggunakan yt_dlp untuk mendownload
        with yt_dlp.YoutubeDL(ydl_opts) as ydl:
            ydl.download([url])

        print("Video dan audio terbaik berhasil didownload dan digabungkan!")

    except Exception as e:
        print(f"Terjadi kesalahan: {e}")

# Ganti dengan URL video YouTube yang ingin didownload
video_url = "https://youtu.be/H_R0dv64zCM?si=6xyfPa0U96ldu9F4"
download_best_quality(video_url)
```

```python
import yt_dlp

def download_audio_as_mp3(url):
    try:
        # Opsi untuk mendownload audio dan konversi ke MP3
        ydl_opts = {
            'format': 'bestaudio/best',           # Hanya ambil audio terbaik
            'outtmpl': '%(title)s.%(ext)s',       # Nama file sesuai judul video
            'noplaylist': True,                   # Hanya satu video, bukan playlist
            'postprocessors': [{
                'key': 'FFmpegExtractAudio',      # Gunakan ffmpeg untuk ekstrak audio
                'preferredcodec': 'mp3',          # Format hasil: MP3
                'preferredquality': '192',        # Kualitas: 192 kbps
            }],
        }

        with yt_dlp.YoutubeDL(ydl_opts) as ydl:
            ydl.download([url])

        print("Audio berhasil didownload dan dikonversi menjadi MP3!")

    except Exception as e:
        print(f"Terjadi kesalahan: {e}")

# Contoh URL video YouTube
video_url = "https://youtu.be/H_R0dv64zCM?si=0u6Gtqe164gJ24XU"
download_audio_as_mp3(video_url)
```

```python
import yt_dlp

def download_video_480p(url):
    try:
        # Opsi untuk mendownload video dengan resolusi maksimal 480p + audio terbaik
        ydl_opts = {
            'format': 'bestvideo[height<=480]+bestaudio/best[height<=480]/best[height<=480]',
            'outtmpl': '%(title)s.%(ext)s',        # Nama file sesuai judul video
            'noplaylist': True,                    # Jangan download playlist
            'merge_output_format': 'mp4',          # Gabungkan hasil ke format mp4
        }

        with yt_dlp.YoutubeDL(ydl_opts) as ydl:
            ydl.download([url])

        print("Video 480p berhasil didownload dan digabungkan sebagai MP4!")

    except Exception as e:
        print(f"Terjadi kesalahan: {e}")

# Ganti dengan URL video YouTube yang ingin didownload
video_url = "https://youtu.be/H_R0dv64zCM?si=0u6Gtqe164gJ24XU"
download_video_480p(video_url)
```
